```python
# %% [markdown]
# # Task 1: Exploratory Data Analysis (EDA)
# ## Insurance Risk Analytics Project

# %%
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# Set style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

# %%
# Load the data
try:
    # Try loading from local path
    df = pd.read_csv('data/raw/insurance_data.csv')
except:
    # If not available, create sample data structure
    print("Please download the data from the provided link")
    # Create sample data for demonstration
    np.random.seed(42)
    n_samples = 10000
    
    data = {
        'PolicyID': np.arange(1, n_samples + 1),
        'TransactionMonth': pd.date_range('2014-02-01', periods=19, freq='M').tolist() * (n_samples // 19),
        'Province': np.random.choice(['Gauteng', 'Western Cape', 'KwaZulu-Natal', 'Eastern Cape', 'Free State'], n_samples),
        'PostalCode': np.random.randint(1000, 9999, n_samples),
        'Gender': np.random.choice(['Male', 'Female'], n_samples, p=[0.55, 0.45]),
        'VehicleType': np.random.choice(['Sedan', 'SUV', 'Truck', 'Hatchback', 'Coupe'], n_samples),
        'Make': np.random.choice(['Toyota', 'Ford', 'BMW', 'Mercedes', 'VW'], n_samples),
        'Model': np.random.choice(['Model_A', 'Model_B', 'Model_C', 'Model_D'], n_samples),
        'RegistrationYear': np.random.randint(2000, 2023, n_samples),
        'SumInsured': np.random.lognormal(10, 1, n_samples),
        'TotalPremium': np.random.lognormal(8, 0.5, n_samples),
        'TotalClaims': np.random.exponential(5000, n_samples) * np.random.binomial(1, 0.1, n_samples),
        'CalculatedPremiumPerTerm': np.random.lognormal(8, 0.3, n_samples),
        'VehicleAge': 2023 - np.random.randint(2000, 2023, n_samples)
    }
    
    df = pd.DataFrame(data)
    df['TransactionMonth'] = pd.to_datetime(df['TransactionMonth'])
    
# %%
# Data Understanding
print("Dataset Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nData Types:")
print(df.dtypes)
print("\nMissing Values:")
print(df.isnull().sum())

# %%
# Descriptive Statistics
print("Descriptive Statistics for Numerical Features:")
print(df.describe())

# %%
# Calculate Loss Ratio
df['LossRatio'] = df['TotalClaims'] / df['TotalPremium']
df['HasClaim'] = (df['TotalClaims'] > 0).astype(int)

print(f"Overall Portfolio Loss Ratio: {df['LossRatio'].mean():.2%}")
print(f"Claim Frequency: {df['HasClaim'].mean():.2%}")

# %%
# Loss Ratio by Province
plt.figure(figsize=(12, 6))
province_loss = df.groupby('Province')['LossRatio'].mean().sort_values()
province_loss.plot(kind='bar', color='skyblue')
plt.title('Loss Ratio by Province', fontsize=16)
plt.xlabel('Province', fontsize=12)
plt.ylabel('Loss Ratio', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('reports/figures/loss_ratio_by_province.png', dpi=300, bbox_inches='tight')
plt.show()

# %%
# Loss Ratio by Gender
gender_loss = df.groupby('Gender')['LossRatio'].mean()
print("Loss Ratio by Gender:")
print(gender_loss)

plt.figure(figsize=(8, 6))
gender_loss.plot(kind='bar', color=['blue', 'pink'])
plt.title('Loss Ratio by Gender', fontsize=16)
plt.xlabel('Gender', fontsize=12)
plt.ylabel('Loss Ratio', fontsize=12)
plt.tight_layout()
plt.savefig('reports/figures/loss_ratio_by_gender.png', dpi=300, bbox_inches='tight')
plt.show()

# %%
# Vehicle Analysis
vehicle_analysis = df.groupby('Make').agg({
    'TotalClaims': 'mean',
    'TotalPremium': 'mean',
    'HasClaim': 'mean',
    'PolicyID': 'count'
}).rename(columns={'PolicyID': 'PolicyCount'})

vehicle_analysis['LossRatio'] = vehicle_analysis['TotalClaims'] / vehicle_analysis['TotalPremium']
vehicle_analysis = vehicle_analysis.sort_values('LossRatio', ascending=False)

print("Top 5 Highest Risk Vehicle Makes:")
print(vehicle_analysis.head())

# %%
# Temporal Trends
monthly_trends = df.groupby(df['TransactionMonth'].dt.to_period('M')).agg({
    'TotalPremium': 'sum',
    'TotalClaims': 'sum',
    'HasClaim': 'mean',
    'PolicyID': 'count'
}).rename(columns={'PolicyID': 'PolicyCount'})

monthly_trends['LossRatio'] = monthly_trends['TotalClaims'] / monthly_trends['TotalPremium']

plt.figure(figsize=(14, 8))
plt.subplot(2, 1, 1)
monthly_trends['LossRatio'].plot(marker='o', color='red')
plt.title('Monthly Loss Ratio Trend', fontsize=16)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Loss Ratio', fontsize=12)
plt.grid(True, alpha=0.3)

plt.subplot(2, 1, 2)
monthly_trends['PolicyCount'].plot(marker='o', color='green')
plt.title('Monthly Policy Count', fontsize=16)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Number of Policies', fontsize=12)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('reports/figures/monthly_trends.png', dpi=300, bbox_inches='tight')
plt.show()

# %%
# Creative Visualization 1: Risk Heatmap by Province and Vehicle Type
plt.figure(figsize=(14, 10))
pivot_data = df.pivot_table(values='LossRatio', 
                           index='Province', 
                           columns='VehicleType', 
                           aggfunc='mean')

sns.heatmap(pivot_data, annot=True, fmt='.2%', cmap='RdYlGn_r', center=0.1)
plt.title('Risk Heatmap: Loss Ratio by Province and Vehicle Type', fontsize=16)
plt.tight_layout()
plt.savefig('reports/figures/risk_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()

# %%
# Creative Visualization 2: Premium vs Claims Distribution
plt.figure(figsize=(12, 10))
plt.subplot(2, 2, 1)
sns.histplot(df['TotalPremium'], kde=True, bins=50, color='blue')
plt.title('Distribution of Total Premium', fontsize=14)
plt.xlabel('Total Premium (ZAR)')

plt.subplot(2, 2, 2)
sns.histplot(df[df['TotalClaims'] > 0]['TotalClaims'], kde=True, bins=50, color='red')
plt.title('Distribution of Claims (Claimants Only)', fontsize=14)
plt.xlabel('Claim Amount (ZAR)')

plt.subplot(2, 2, 3)
sns.scatterplot(x='VehicleAge', y='TotalClaims', data=df[df['TotalClaims'] > 0], alpha=0.6)
plt.title('Claims vs Vehicle Age', fontsize=14)
plt.xlabel('Vehicle Age (Years)')
plt.ylabel('Claim Amount (ZAR)')

plt.subplot(2, 2, 4)
sns.boxplot(x='Gender', y='LossRatio', data=df[df['LossRatio'].notna() & (df['LossRatio'] < 2)])
plt.title('Loss Ratio Distribution by Gender', fontsize=14)
plt.xlabel('Gender')
plt.ylabel('Loss Ratio')

plt.tight_layout()
plt.savefig('reports/figures/distribution_analysis.png', dpi=300, bbox_inches='tight')
plt.show()

# %%
# Creative Visualization 3: Interactive Correlation Matrix
numeric_cols = ['TotalPremium', 'TotalClaims', 'SumInsured', 'VehicleAge', 'LossRatio']
corr_matrix = df[numeric_cols].corr()

plt.figure(figsize=(10, 8))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(corr_matrix, mask=mask, annot=True, cmap='coolwarm', center=0,
            square=True, linewidths=1, cbar_kws={"shrink": .8})
plt.title('Correlation Matrix of Key Variables', fontsize=16)
plt.tight_layout()
plt.savefig('reports/figures/correlation_matrix.png', dpi=300, bbox_inches='tight')
plt.show()

# %%
# Outlier Detection
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
sns.boxplot(y=df['TotalPremium'])
plt.title('Total Premium Outliers')

plt.subplot(1, 3, 2)
sns.boxplot(y=df[df['TotalClaims'] > 0]['TotalClaims'])
plt.title('Claim Amount Outliers (Claimants Only)')

plt.subplot(1, 3, 3)
sns.boxplot(y=df[df['LossRatio'].notna() & (df['LossRatio'] < 3)]['LossRatio'])
plt.title('Loss Ratio Outliers')

plt.tight_layout()
plt.savefig('reports/figures/outlier_detection.png', dpi=300, bbox_inches='tight')
plt.show()

# %%
# Key Insights Summary
print("="*60)
print("KEY INSIGHTS FROM EDA")
print("="*60)

print(f"\n1. Overall Portfolio Metrics:")
print(f"   - Total Policies: {len(df):,}")
print(f"   - Claim Frequency: {df['HasClaim'].mean():.2%}")
print(f"   - Average Loss Ratio: {df['LossRatio'].mean():.2%}")

print(f"\n2. Risk Variation by Province:")
for province, loss_ratio in province_loss.items():
    print(f"   - {province}: {loss_ratio:.2%}")

print(f"\n3. Gender Differences:")
print(f"   - Male Loss Ratio: {gender_loss.get('Male', 0):.2%}")
print(f"   - Female Loss Ratio: {gender_loss.get('Female', 0):.2%}")

print(f"\n4. Temporal Trends:")
print(f"   - Period Covered: {df['TransactionMonth'].min().date()} to {df['TransactionMonth'].max().date()}")
print(f"   - Monthly Policy Growth: {((monthly_trends['PolicyCount'].iloc[-1] / monthly_trends['PolicyCount'].iloc[0]) - 1):.2%}")

print(f"\n5. Data Quality:")
print(f"   - Missing Values: {df.isnull().sum().sum()} total")
print(f"   - Numerical Outliers: Found in Premium, Claims, and Loss Ratio")
```
