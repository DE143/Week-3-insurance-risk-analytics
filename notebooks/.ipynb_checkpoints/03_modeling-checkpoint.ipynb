```python
# %% [markdown]
# # Task 4: Predictive Modeling
# ## Machine Learning for Risk Assessment and Premium Optimization

# %%
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.preprocessing import StandardScaler, OneHotEncoder, LabelEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error, classification_report, roc_auc_score
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from xgboost import XGBRegressor, XGBClassifier
import lightgbm as lgb
import catboost as cb
import shap
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import json
from datetime import datetime

# %%
# Load and prepare data
df = pd.read_csv('data/processed/insurance_processed.csv')

# Feature engineering
df['VehicleAge'] = 2023 - df['RegistrationYear']
df['PremiumPerInsured'] = df['TotalPremium'] / df['SumInsured']
df['ClaimRatio'] = df['TotalClaims'] / df['SumInsured']
df['HasClaim'] = (df['TotalClaims'] > 0).astype(int)

# Select relevant features
features = [
    # Client features
    'Gender', 'MaritalStatus', 'Province', 'PostalCode',
    
    # Vehicle features
    'VehicleType', 'Make', 'Model', 'VehicleAge', 'Cylinders',
    'Cubiccapacity', 'Kilowatts', 'Bodytype', 'NumberOfDoors',
    
    # Insurance features
    'SumInsured', 'TermFrequency', 'ExcessSelected',
    'CoverType', 'Product', 'StatutoryClass'
]

target_claim = 'TotalClaims'
target_premium = 'TotalPremium'
target_claim_flag = 'HasClaim'

# Prepare data
X = df[features].copy()
y_claim = df[target_claim]
y_premium = df[target_premium]
y_claim_flag = df[target_claim_flag]

# %%
# Data preprocessing
print("Data Preprocessing...")

# Identify categorical and numerical columns
categorical_cols = X.select_dtypes(include=['object', 'category']).columns.tolist()
numerical_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()

print(f"Categorical features ({len(categorical_cols)}): {categorical_cols}")
print(f"Numerical features ({len(numerical_cols)}): {numerical_cols}")

# Create preprocessing pipelines
numerical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

# Combine transformers
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_cols),
        ('cat', categorical_transformer, categorical_cols)
    ])

# %%
# Model 1: Claim Severity Prediction (Regression)
print("\n" + "="*60)
print("MODEL 1: CLAIM SEVERITY PREDICTION")
print("="*60)

# Filter only claims data
claim_data = df[df['TotalClaims'] > 0].copy()
X_claim = claim_data[features]
y_claim_actual = claim_data[target_claim]

# Split data
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_claim, y_claim_actual, test_size=0.2, random_state=42
)

# Define models
models = {
    'Linear Regression': LinearRegression(),
    'Random Forest': RandomForestRegressor(n_estimators=100, random_state=42),
    'XGBoost': XGBRegressor(n_estimators=100, random_state=42, verbosity=0),
    'LightGBM': lgb.LGBMRegressor(n_estimators=100, random_state=42, verbose=-1)
}

# Evaluate models
results = {}
for name, model in models.items():
    print(f"\nTraining {name}...")
    
    # Create pipeline
    pipeline = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('regressor', model)
    ])
    
    # Train model
    pipeline.fit(X_train_c, y_train_c)
    
    # Predict
    y_pred = pipeline.predict(X_test_c)
    
    # Calculate metrics
    rmse = np.sqrt(mean_squared_error(y_test_c, y_pred))
    mae = mean_absolute_error(y_test_c, y_pred)
    r2 = r2_score(y_test_c, y_pred)
    
    results[name] = {
        'RMSE': rmse,
        'MAE': mae,
        'R2': r2,
        'model': pipeline
    }
    
    print(f"  RMSE: {rmse:,.2f}")
    print(f"  MAE: {mae:,.2f}")
    print(f"  R²: {r2:.4f}")

# %%
# Model 2: Claim Probability Prediction (Classification)
print("\n" + "="*60)
print("MODEL 2: CLAIM PROBABILITY PREDICTION")
print("="*60)

# Split data for classification
X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(
    X, y_claim_flag, test_size=0.2, random_state=42, stratify=y_claim_flag
)

# Define classification models
clf_models = {
    'Logistic Regression': LogisticRegression(random_state=42, max_iter=1000),
    'Random Forest Classifier': RandomForestClassifier(n_estimators=100, random_state=42),
    'XGBoost Classifier': XGBClassifier(n_estimators=100, random_state=42, verbosity=0),
    'LightGBM Classifier': lgb.LGBMClassifier(n_estimators=100, random_state=42, verbose=-1)
}

clf_results = {}
for name, model in clf_models.items():
    print(f"\nTraining {name}...")
    
    # Create pipeline
    pipeline = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])
    
    # Train model
    pipeline.fit(X_train_clf, y_train_clf)
    
    # Predict
    y_pred_proba = pipeline.predict_proba(X_test_clf)[:, 1]
    y_pred = pipeline.predict(X_test_clf)
    
    # Calculate metrics
    roc_auc = roc_auc_score(y_test_clf, y_pred_proba)
    
    clf_results[name] = {
        'ROC_AUC': roc_auc,
        'model': pipeline,
        'report': classification_report(y_test_clf, y_pred, output_dict=True)
    }
    
    print(f"  ROC AUC: {roc_auc:.4f}")

# %%
# Model 3: Premium Optimization Model
print("\n" + "="*60)
print("MODEL 3: PREMIUM OPTIMIZATION")
print("="*60)

# Calculate optimal premium (simplified version)
# In reality, this would consider risk, expenses, profit margin, competition
df['OptimalPremium'] = df.apply(
    lambda row: row['CalculatedPremiumPerTerm'] * (1 + 0.1 * row['HasClaim']) * (1 + 0.05 * (row['VehicleAge'] > 10)),
    axis=1
)

X_premium = df[features]
y_optimal = df['OptimalPremium']

# Split data
X_train_p, X_test_p, y_train_p, y_test_p = train_test_split(
    X_premium, y_optimal, test_size=0.2, random_state=42
)

# Train premium optimization model
premium_models = {
    'XGBoost Premium': XGBRegressor(n_estimators=150, random_state=42, verbosity=0),
    'LightGBM Premium': lgb.LGBMRegressor(n_estimators=150, random_state=42, verbose=-1)
}

premium_results = {}
for name, model in premium_models.items():
    print(f"\nTraining {name}...")
    
    pipeline = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('regressor', model)
    ])
    
    # Hyperparameter tuning
    param_grid = {
        'regressor__max_depth': [3, 5, 7],
        'regressor__learning_rate': [0.01, 0.1, 0.2],
        'regressor__subsample': [0.8, 0.9, 1.0]
    }
    
    grid_search = GridSearchCV(
        pipeline, param_grid, cv=3, scoring='r2', n_jobs=-1, verbose=0
    )
    
    grid_search.fit(X_train_p, y_train_p)
    
    # Best model
    best_model = grid_search.best_estimator_
    y_pred = best_model.predict(X_test_p)
    
    rmse = np.sqrt(mean_squared_error(y_test_p, y_pred))
    r2 = r2_score(y_test_p, y_pred)
    
    premium_results[name] = {
        'RMSE': rmse,
        'R2': r2,
        'best_params': grid_search.best_params_,
        'model': best_model
    }
    
    print(f"  Best R²: {r2:.4f}")
    print(f"  Best RMSE: {rmse:,.2f}")
    print(f"  Best params: {grid_search.best_params_}")

# %%
# Model Comparison and Selection
print("\n" + "="*60)
print("MODEL COMPARISON")
print("="*60)

# Compare claim severity models
print("\nClaim Severity Models (Lower RMSE is better):")
severity_df = pd.DataFrame({
    'Model': list(results.keys()),
    'RMSE': [r['RMSE'] for r in results.values()],
    'R²': [r['R2'] for r in results.values()],
    'MAE': [r['MAE'] for r in results.values()]
}).sort_values('RMSE')

print(severity_df.to_string(index=False))

# Compare classification models
print("\n\nClaim Probability Models (Higher AUC is better):")
clf_df = pd.DataFrame({
    'Model': list(clf_results.keys()),
    'ROC AUC': [r['ROC_AUC'] for r in clf_results.values()]
}).sort_values('ROC AUC', ascending=False)

print(clf_df.to_string(index=False))

# Select best models
best_severity_model = severity_df.iloc[0]['Model']
best_clf_model = clf_df.iloc[0]['Model']

print(f"\nSelected Models:")
print(f"  Best Severity Model: {best_severity_model}")
print(f"  Best Classification Model: {best_clf_model}")

# %%
# SHAP Analysis for Model Interpretation
print("\n" + "="*60)
print("MODEL INTERPRETATION WITH SHAP")
print("="*60)

# Prepare data for SHAP
X_processed = preprocessor.fit_transform(X)
feature_names = (numerical_cols + 
                 list(preprocessor.named_transformers_['cat']
                     .named_steps['onehot']
                     .get_feature_names_out(categorical_cols)))

# Train best model for SHAP analysis
best_model_pipeline = results[best_severity_model]['model']
best_model = best_model_pipeline.named_steps['regressor']

# Get preprocessed data
X_train_processed = best_model_pipeline.named_steps['preprocessor'].transform(X_train_c)

# Create SHAP explainer
print("Calculating SHAP values...")
explainer = shap.TreeExplainer(best_model)
shap_values = explainer.shap_values(X_train_processed)

# Summary plot
plt.figure(figsize=(12, 8))
shap.summary_plot(shap_values, X_train_processed, feature_names=feature_names, 
                  max_display=15, show=False)
plt.title(f'SHAP Summary Plot - {best_severity_model}', fontsize=16)
plt.tight_layout()
plt.savefig('reports/figures/shap_summary.png', dpi=300, bbox_inches='tight')
plt.show()

# Feature importance
shap_importance = pd.DataFrame({
    'feature': feature_names,
    'importance': np.abs(shap_values).mean(axis=0)
}).sort_values('importance', ascending=False).head(10)

print("\nTop 10 Most Important Features:")
print(shap_importance.to_string(index=False))

# Dependence plots for top features
top_features = shap_importance.head(3)['feature'].tolist()
for i, feature in enumerate(top_features[:2]):
    plt.figure(figsize=(10, 6))
    shap.dependence_plot(feature, shap_values, X_train_processed, 
                        feature_names=feature_names, show=False)
    plt.title(f'SHAP Dependence Plot: {feature}', fontsize=14)
    plt.tight_layout()
    plt.savefig(f'reports/figures/shap_dependence_{i}.png', dpi=300, bbox_inches='tight')
    plt.show()

# %%
# Business Application: Risk-Based Pricing Framework
print("\n" + "="*60)
print("RISK-BASED PRICING FRAMEWORK")
print("="*60)

def calculate_risk_based_premium(row, clf_model, severity_model, preprocessor, 
                                 expense_loading=0.15, profit_margin=0.10):
    """
    Calculate risk-based premium using predicted probability and severity
    """
    # Prepare input
    input_data = pd.DataFrame([row])
    
    # Predict claim probability
    claim_prob = clf_model.predict_proba(input_data)[0, 1]
    
    # Predict claim severity (if claim occurs)
    predicted_severity = severity_model.predict(input_data)[0]
    
    # Calculate pure premium
    pure_premium = claim_prob * predicted_severity
    
    # Add loadings
    risk_based_premium = pure_premium * (1 + expense_loading) * (1 + profit_margin)
    
    return {
        'claim_probability': claim_prob,
        'predicted_severity': predicted_severity,
        'pure_premium': pure_premium,
        'risk_based_premium': risk_based_premium,
        'current_premium': row['CalculatedPremiumPerTerm']
    }

# Test on sample data
sample_indices = [0, 100, 500, 1000]
print("\nSample Risk-Based Premium Calculations:")
print("-" * 80)

for idx in sample_indices:
    if idx < len(df):
        sample_row = df.iloc[idx]
        result = calculate_risk_based_premium(
            sample_row,
            clf_results[best_clf_model]['model'],
            results[best_severity_model]['model'],
            preprocessor
        )
        
        print(f"\nPolicy {idx}:")
        print(f"  Current Premium: ZAR {result['current_premium']:,.2f}")
        print(f"  Risk-Based Premium: ZAR {result['risk_based_premium']:,.2f}")
        print(f"  Difference: ZAR {result['risk_based_premium'] - result['current_premium']:,.2f}")
        print(f"  Claim Probability: {result['claim_probability']:.2%}")
        print(f"  Predicted Severity: ZAR {result['predicted_severity']:,.2f}")

# %%
# Model Deployment Preparation
print("\n" + "="*60)
print("MODEL DEPLOYMENT PREPARATION")
print("="*60)

# Save models
model_artifacts = {
    'preprocessor': preprocessor,
    'severity_model': results[best_severity_model]['model'],
    'classification_model': clf_results[best_clf_model]['model'],
    'premium_model': premium_results['XGBoost Premium']['model'],
    'metadata': {
        'created_date': datetime.now().isoformat(),
        'features': features,
        'best_severity_model': best_severity_model,
        'best_classification_model': best_clf_model,
        'model_metrics': {
            'severity_rmse': results[best_severity_model]['RMSE'],
            'severity_r2': results[best_severity_model]['R2'],
            'clf_auc': clf_results[best_clf_model]['ROC_AUC']
        }
    }
}

# Create models directory
import os
os.makedirs('models', exist_ok=True)

# Save artifacts
joblib.dump(model_artifacts, 'models/model_artifacts.pkl')
print("Models saved to 'models/model_artifacts.pkl'")

# Save metrics
metrics = {
    'claim_severity_models': severity_df.to_dict('records'),
    'claim_probability_models': clf_df.to_dict('records'),
    'shap_importance': shap_importance.to_dict('records')
}

with open('models/metrics.json', 'w') as f:
    json.dump(metrics, f, indent=2)
print("Metrics saved to 'models/metrics.json'")

# %%
# Final Recommendations
print("\n" + "="*60)
print("FINAL RECOMMENDATIONS")
print("="*60)

print("\n1. MODEL SELECTION:")
print(f"   • Use {best_severity_model} for claim severity prediction")
print(f"   • Use {best_clf_model} for claim probability prediction")
print(f"   • Expected prediction error: ZAR {results[best_severity_model]['RMSE']:,.2f} RMSE")

print("\n2. KEY RISK DRIVERS (from SHAP analysis):")
for i, (feature, importance) in enumerate(zip(shap_importance['feature'], shap_importance['importance']), 1):
    print(f"   {i}. {feature}: {importance:.4f}")

print("\n3. PRICING STRATEGY:")
print("   • Implement risk-based pricing using predicted probability × severity")
print("   • Adjust premiums by ±20% based on risk score")
print("   • Create discount tiers for low-risk segments")

print("\n4. MARKETING RECOMMENDATIONS:")
print("   • Target low-risk zip codes identified in hypothesis testing")
print("   • Focus on vehicle makes with lower claim frequencies")
print("   • Offer bundled products for high-value customers")

print("\n5. IMPLEMENTATION ROADMAP:")
print("   • Month 1-2: Pilot risk-based pricing in test regions")
print("   • Month 3-4: Full model deployment with monitoring")
print("   • Month 5-6: Continuous improvement with new data")

print("\n6. RISK MITIGATION:")
print("   • Regular model retraining (quarterly)")
print("   • Fairness testing across protected attributes")
print("   • Regulatory compliance review")
```
